```python
# 2D array
climate_data = np.array([[73, 67, 43],
                         [91, 8, 64],
                         [87, 134, 58],
                         [102, 43, 37],
                         [69, 96, 70]])
```

```python
# 2D array (matrix)
# We can inspect the length along each dimension using the .shape property of an array.
# it gives a tuple
climate_data.shape 
```

```python
# dtype -> tells data type
climate_data.dtype
```

All the elements in numpy array has same datatype. if even a single integer is coverted to float then all will be converted into float or string.

<U21 means: A Unicode string array where each element can hold up to 21 characters.

```python
weight = np.array([0.3, 0.2, 0.5])
```

```python
# matmul -> function for matrix multiplication in numpy
np.matmul(climate_data,weight)
```

```python
climate_data @ weight # another method for matrix multiplication in numpy
```
