```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
y = [10,50,100]
x = [1,5,9]
```

```python
plt.plot(x,y,marker='x',ls='-',c='r');
plt.plot(y,x,'o-b');
plt.xlabel('X value');
plt.ylabel('Y value');
plt.title('Graph')
plt.legend(['Y vs X','X vs Y'])
```

```python
flower_df=sns.load_dataset('iris')
```

```python
flower_df
```

```python
plt.plot(flower_df.sepal_length,flower_df.sepal_width,'o')
```

```python
sns.scatterplot(x=flower_df.sepal_length, y=flower_df.sepal_width);
```

```python
plt.figure(figsize=(12,6))
plt.title("Sepal Plot")
sns.scatterplot(data=flower_df,x='sepal_length',y='sepal_width',hue='species',s=100)
```

```python
sns.histplot(data=flower_df,x='sepal_width',bins=10,alpha=0.9);
```

```python
plt.hist(flower_df.sepal_width);
```

```python
import numpy as np

# Specifying the boundaries of each bin
plt.hist(flower_df.sepal_width, bins=np.arange(2, 5, 0.25));
```
