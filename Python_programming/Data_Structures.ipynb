There are Four Types of Data Structures in Python:-
1: list
2: tuple
3: set
4: Dictionary

list -> Ordered, Mutable, Duplicates
tuple -> Ordered, Immutable, Duplicates
Set -> Unordered, Addable/Removable, no Duplicates
Dictionary -> Unordered, Changeable, no Duplicates

Ordered: Can be accessed using index

```python
# Declaration
L = [1, 23, 5, "PP", 3.4]
T = (1, 23, 5, "PP", 3.4)
S = {1, 1000, 5, "aP", 3.4}
D = {1:"One", 23:"TwentyThree", 5:"Five", "PP":"DoubleP", 3.4:3.4000}
```

```python
print(L,T,S,D,sep='\n')
```

```python
# Type
print("The type of L is ", type(L))
print("The type of T is ", type(T))
print("The type of S is ", type(S))
print("The type of D is ", type(D))
```

```python
# Accessing element
print(L[2])
print(T[3])
print(5 in S) # finds the key inside set
print(D["PP"]) # gives value corresponding to that key
```

```python
# list and tuple are same only exception is mutability
print(L[2:5])
print(T[2:5])
```

```python
# Adding element
L = L + ["poopie"] # L.append("poopie") is faster
T3 = T + T # Changing element in a tuple is not allowed
S.add("poopie")
S.update({"ThreePointFour","peckpii"})
D["pew"] = "poopie"
print(L,T3,S,D,sep='\n')
```

```python
# + can't be used to add two dictionaries but we can use update function
D2 = {"ploo":4,3:"plee"}
D.update(D2)
print(D)
```

```python
# Copying
L2 = L.copy() 
T2 = T # value will be assigned only T2 is not declared previously
S2 = S.copy()
D2 = D.copy()
print(L2,T2,S2,D2,sep='\n')
```

```python
"""Let say you are a teacher and you have different student
records containing id of a student and the marks list in each subject
where different students have taken different number of subjects. All
these records are in hard copy. You want to enter all the data in computer
and want to compute the average marks of each student and display"""

def getDataFromUser():
    D = {}
    while True:
        studentId = input("Enter student ID: ")
        marksList = input("Enter the marks by comma separted values: ")
        moreStudents = input('Enter "no" to quit insertion: ')
        if studentId in D:
            print(studentId, "is already insterted")
        else:
            D[studentId] = marksList.split(",")
        if moreStudents.lower() == "no":
            return D
```

```python
studentData = {'12': ['56', '45', '13'],
 '45': ['44', '55', '66', '77', '4'],
 '23': ['45', '45']}
```

```python
studentData
```

```python
def getAvgMarks(l):
    sum = 0
    for j in l:
        sum += int(j)
    n = len(l)
    return sum/n

for i in studentData:
    L = studentData[i]
    print(getAvgMarks(L))
```
