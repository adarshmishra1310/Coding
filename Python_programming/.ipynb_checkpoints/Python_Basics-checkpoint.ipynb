# Loops

```python
# While loop
n = int(input("Enter n: "))
i = 0;
while(i<n):
    print(i+1,end=' ')
    i+=1
print("\ndone")
```

```python
# pass -> do nothing
# break -> exit the loop
# continue -> skips to next iteration
```

```python
# For loop
# if starting range not specified then it starts from 0
for i in range(10):
    print(i + 1, end = ' ')
print('\n')

# runs from i=1 to 10 (10 is excluded)
for i in range (1,10):
    print(i + 1, end = ' ')
print('\n')

# runs from i=1 to 10 and everytime i is incremented by 3(i+=3) 
for i in range (1,10,3):
    print(i + 1,end = ' ')
```

# Some Problems

```python
# CF question
import math

t = int(input())
for i in range(t):
    a,b = map(int, input().split()) # used to take input separately
    print(math.lcm(a,b))
```

```python
# Sort the given array
l = [1,2,4,-5,7,9,3,2]
ans = []
n = len(l)
for i in range(n):
    mn = 0
    sz = len(l)
    for j in range(sz):
        if(l[j] < l[mn]):
            mn = j
    ans.append(l[mn])
    del l[mn]
for i in ans:
    print(i,end = ' ')
```

```python
def printMsg(msg):
    """ The function prints the message supplied by the user
    or prints that msg is not in the form of string"""
    
    if isinstance(msg,str):
        print(msg)
    else:
        print("Your input argument is not a string")
        print("Here is the type of what you have supplied :",type(msg))

y = "This is the message"
printMsg(y)
print(type(printMsg(y)))
```

```python
def myAddUniversal(*args):
    s = 0
    for i in range(len(args)):
        s += args[i] # s = s+args[i]
    return s

print(myAddUniversal(2,4,5,4.6,78)) # for variable number of arguements
```

```python
def printAllVariableNamesAndValues(**args):
    for x in args:
        print("Variable Name is :",x," And Value is :",args[x])

printAllVariableNamesAndValues(a = 3,b="B",c="CCC",y=6.7) # for variable number of arguements with variables itself
```

```python
# pass a function as a arguement
def mult_by_five(x):
    return 5 * x

def call(fn, arg):
    """Call fn on arg"""
    return fn(arg)

def squared_call(fn, arg):
    """Call fn on the result of calling fn on arg"""
    return fn(fn(arg))

print(
    call(mult_by_five, 1),
    squared_call(mult_by_five, 1), 
    sep='\n', # '\n' is the newline character - it starts a new line
)
```

```python
max?
```

a function can also be passed with max or min. It executes the function first and then tell max/min of that returned value.

```python
def mod_5(x):
    """Return the remainder of x after dividing by 5"""
    return x % 5

print(
    'Which number is biggest?',
    max(100, 51, 14),
    'Which number is the biggest modulo 5?',
    max(100, 51, 14, key=mod_5),
    sep='\n',
)
```
